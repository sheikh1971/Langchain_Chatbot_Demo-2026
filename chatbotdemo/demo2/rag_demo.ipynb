```python
#web based loader
from langchain_community.document_loaders import WebBaseLoader
import bs4

##load,chunk and index the data
loader = WebBaseLoader(web_paths=["https://activisthandbook.org/theory/what-is-activism?gad_source=1&gad_campaignid=16953797910&gbraid=0AAAAAojphQQUO74toNHL8JgzdS-xdcMvC&gclid=CjwKCAjw5ZXQBhBdEiwAI5XVWYJl4-woFe_LlHGpbVHd65c0opImMGlX_kPlUDA-lKJmnLfrMD07ixoCVlIQAvD_BwEttps://en.wikipedia.org/wiki/air_pollution"])

bs_kwargs = dict ( parse_only= bs4.SoupStrainer(class_=("description","meta")) )

text_documents = loader.load()
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.vectorstores import Chroma
from langchain_community.document_loaders import PyPDFLoader
from langchain_nvidia_ai_endpoints import NVIDIAEmbeddings
import os
from dotenv import load_dotenv
load_dotenv()   
#importing getenv to access environment variables
from os import getenv
# Load PDF
pdf_loader = PyPDFLoader("E:\\Langchain Project\\Chat_Model\\chatbotdemo\\public\\AI-Based_Mental_Health_Chatbot_using_LangChain_and.pdf")
pdf_documents = pdf_loader.load()

# Combine all documents
all_documents = documents + text_documents + pdf_documents

# Split text into chunks
text_splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=200)
chunks = text_splitter.split_documents(all_documents)

# Create NVIDIA embeddings and vector store
embeddings = NVIDIAEmbeddings(
    model="nvidia/nv-embed-v1",
    api_key= getenv("NVIDIA_API_KEY"),    
          # Replace with your actual NVIDIA API key
    truncate="NONE"
    )
vector_store = Chroma.from_documents(documents=chunks, embedding=embeddings, persist_directory="./chroma_db")

print(f"Loaded {len(chunks)} document chunks into Chroma vector store")
```

```python
results = vector_store.similarity_search("What is the main objective of this paper?", k=2)
for doc in results:
    print(doc.page_content)
```

```python
from langchain_text_splitters import TokenTextSplitter
from langchain_community.vectorstores import Chroma
from langchain_community.document_loaders import PyPDFLoader
from langchain_nvidia_ai_endpoints import NVIDIAEmbeddings
import os
from dotenv import load_dotenv
load_dotenv()
from os import getenv
# Load PDF
pdf_loader = PyPDFLoader("E:\\Langchain Project\\Chat_Model\\chatbotdemo\\public\\AI-Based_Mental_Health_Chatbot_using_LangChain_and.pdf")
pdf_documents = pdf_loader.load()

# Combine all documents
all_documents = documents + text_documents + pdf_documents

# Split text into chunks using TokenTextSplitter for more reliable, token-based splitting
text_splitter = TokenTextSplitter(chunk_size=256, chunk_overlap=32)
chunks = text_splitter.split_documents(all_documents)

# Create NVIDIA embeddings and vector store
embeddings = NVIDIAEmbeddings(
    model="nvidia/nv-embed-v1",
    api_key= getenv("NVIDIA_API_KEY"),
    truncate="NONE"
    )
vector_store = Chroma.from_documents(documents=chunks, embedding=embeddings, persist_directory="./chroma_db")

print(f"Loaded {len(chunks)} document chunks into Chroma vector store")

results = vector_store.similarity_search("What is the main objective of this paper?", k=2)
for doc in results:
    print(doc.page_content)
```
